```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("polynomial_data.csv")
```

```python
X = data["x_train"].values
y = data["y_train"].values

x_train = X.reshape(-1,1)
y_train = y.reshape(-1,1)

# Plotting the data
plt.figure(figsize = (8, 5))
plt.scatter(x_train, y_train, color = "blue", label = "Training data", alpha = 0.7)
plt.title("Dummy Linear Regression Data")
plt.xlabel("x_train")
plt.ylabel("y_train")
plt.legend()
plt.grid(True)
plt.show()

print("Features shape is:", x_train.shape)
print("Target shape is:", y_train.shape)
```

```python
m = x_train.shape[0]   
d = 17

x = np.zeros((m,d))

for i in range(1,d+1):
    x[:,i-1] = x_train[:,0]**i

mean = np.mean(x,axis=0)
std = np.std(x,axis=0)

x=(x-mean)/std

alpha = 0.1

w = np.random.randn(1,x.shape[1])
b = np.array([[0]])


cost_history = []
for _ in range(10000):
    y_ = (w @ x.T + b).T
    
    cost = np.sum((y_ - y_train) ** 2) / m
    cost_history.append(cost)
   
    
    djdw =(x.T @ (y_ - y_train)).T / m
    djdb = np.sum(y_ - y_train) / m
    
    w = w - alpha * djdw
    b = b - alpha * djdb
y_ = (w @ x.T+b).T
print(y_.shape)
    
    
# Plotting the data
plt.figure(figsize = (8, 5))
plt.scatter(x_train, y_train, color = "blue", label = "Training data", alpha = 0.7)
plt.plot(x_train, y_)
plt.title("Dummy Linear Regression Data")
plt.xlabel("x_train")
plt.ylabel("y_train")
plt.legend()
plt.grid(True)
plt.show() 
plt.plot(cost_history)
plt.show()


                   
```
